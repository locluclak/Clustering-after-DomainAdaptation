```python
df = pd.read_csv('rawdata.csv')
```

```python
df.head()
```

```python
features = df.columns.tolist()
features
```

```python
df = pd.get_dummies(df, columns=[
    "MTRANS"
])
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

target = df['NObeyesdad']
df.drop(columns=['NObeyesdad'],inplace=True)
le = LabelEncoder()
target = le.fit_transform(target)
```

```python
features = df.columns.tolist()
features
```

```python
ordinal_mapping = {'no': 0, 'Sometimes': 1, 'Frequently': 2, 'Always': 3}
df['CAEC'] = df['CAEC'].map(ordinal_mapping)
df['CALC'] = df['CALC'].map(ordinal_mapping)
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_cols = ['Gender', 'family_history_with_overweight', 'FAVC', 'SMOKE', 'SCC']
encoder = OrdinalEncoder()
df[ordinal_cols] = encoder.fit_transform(df[ordinal_cols])
```

```python
from sklearn.preprocessing import StandardScaler

scale_cols = ['Age', 'Height', 'Weight', 'CH2O', 'FAF', 'CALC', "FCVC", "NCP"]
scaler = StandardScaler()
df[scale_cols] = scaler.fit_transform(df[scale_cols])
df.head()
```

```python
df['NObeyesdad'] = target
```

```python
df_gender_0 = df[df['Gender'] == 0.0].copy()
df_gender_1 = df[df['Gender'] == 1.0].copy()
```

```python
df_gender_0.drop(columns=['Gender'], inplace=True)
df_gender_1.drop(columns=['Gender'], inplace=True)
```
