```python
with open("obesity_pvalue_oc.txt", "r") as f:
    pvalues_oc = [
        float(line.strip()) 
        for line in f 
        if line.strip() and line.strip() != "None"
    ]
```

```python
with open("obesity_pvalue_para.txt", "r") as f:
    pvalues_para = [
        float(line.strip()) 
        for line in f 
        if line.strip() and line.strip() != "None"
    ]
```

```python
with open("obesity_pvalue_permutation.txt", "r") as f:
    pvalues_perm = [
        float(line.strip()) 
        for line in f 
        if line.strip() and line.strip() != "None"
    ]
```

```python
import matplotlib.pyplot as plt

data = [pvalues_oc, pvalues_para, pvalues_perm]

plt.figure(figsize=(8, 6))
plt.boxplot(data, labels=['pvalues_oc', 'pvalues_para', 'pvalues_perm'])
plt.ylabel('p-value')
plt.title('Boxplot of p-value ')
plt.ylim(0, 1)
plt.show()
```
